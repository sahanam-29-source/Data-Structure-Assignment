```python
for i in range(1,554):
    response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={}'.format(i))
    data = response.json()
    if 'results' not in data or not data['results']:
        print(f"No more results or error for page {i}. Continue loop.")
        continue
    temp_df = pd.DataFrame(data['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
    df = pd.concat([df, temp_df], ignore_index=True)
```

Question 20 – Calculate the profit for each movie and find the top 10 movies by estimated profit.

```python
df["profit"] = df["grossWorldwide"] - df["budget"]

top10_profit = df.sort_values("profit", ascending=False).head(10)

print(top10_profit[["primaryTitle", "budget", "grossWorldwide", "profit"]])
```

Question 19 – Find the top 10 movies based on the grossWorldwide/budget ratio.

```python
movies = df.dropna(subset=["budget", "grossWorldwide"])

movies["ratio"] = movies["grossWorldwide"] / movies["budget"]

top10_ratio = movies.sort_values("ratio", ascending=False).head(10)

print(top10_ratio[["primaryTitle", "budget", "grossWorldwide", "ratio"]])
```
