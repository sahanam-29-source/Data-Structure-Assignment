````python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Website URL
page_url = "https://www.cardekho.com/used-cars+in+chennai/page-2"

# 2. Send request
response = requests.get(
    page_url,
    headers={"User-Agent": "Mozilla/5.0"}
)

print("Status Code:", response.status_code)

# 3. Create BeautifulSoup object
soup = BeautifulSoup(response.content, "html.parser")

# 4. Find all car cards
cars = soup.find_all(
    "div",
    class_="NewUcExCard posR moreHeight"
)

print("Number of cars found:", len(cars))


# 5. List to store scraped data
data = []


# 6. Scrape each car
for car in cars:

    # -----------------------------
    # CAR NAME + YEAR
    # -----------------------------

    car_link = car.find("h3", class_="title")

    if car_link:
        a_tag = car_link.find("a")

        if a_tag:
            car_name = a_tag.get_text(" ", strip=True)

            # Get year from the span
            year_tag = a_tag.find("span")

            if year_tag:
                manufacturing_year = year_tag.get_text(strip=True)
            else:
                manufacturing_year = "Not available"

        else:
            car_name = "Not available"
            manufacturing_year = "Not available"

    else:
        car_name = "Not available"
        manufacturing_year = "Not available"


    # -----------------------------
    # CAR MAKE
    # -----------------------------

    # Common car manufacturers
    car_makes = [
        "Maruti Suzuki",
        "Mahindra",
        "Land Rover",
        "Mercedes-Benz",
        "Rolls-Royce",
        "Volkswagen",
        "Renault",
        "Hyundai",
        "Toyota",
        "Honda",
        "Skoda",
        "Nissan",
        "Kia",
        "Tata",
        "Ford",
        "BMW",
        "Audi",
        "Volvo",
        "Jeep",
        "MG",
        "Citroen",
        "Datsun",
        "Fiat",
        "Jaguar",
        "Porsche",
        "Lexus",
        "Isuzu",
        "Mitsubishi"
    ]

    car_make = "Not available"

    for make in car_makes:
        if car_name.lower().startswith(make.lower()):
            car_make = make
            break


    # -----------------------------
    # CAR MODEL
    # -----------------------------

    if car_make != "Not available":
        car_model = car_name[len(car_make):].strip()
    else:
        car_model = car_name


    # -----------------------------
    # PRICE
    # -----------------------------

    price_tag = car.find("div", class_="Price")

    if price_tag:
        price_p = price_tag.find("p")

        if price_p:
            price = price_p.get_text(strip=True)
        else:
            price = "Not available"
    else:
        price = "Not available"


    # -----------------------------
    # KILOMETERS, TRANSMISSION, FUEL
    # -----------------------------

    details = car.find("div", class_="dotsDetails")

    kilometers_driven = "Not available"
    transmission = "Not available"
    fuel_type = "Not available"

    if details:

        detail_divs = details.find_all("div")

        values = [
            d.get_text(" ", strip=True)
            for d in detail_divs
        ]

        if len(values) >= 1:
            kilometers_driven = values[0]

        if len(values) >= 2:
            transmission = values[1]

        if len(values) >= 3:
            fuel_type = values[2]


    # -----------------------------
    # LOCATION + CITY
    # -----------------------------

    location_tag = car.find("div", class_="distanceText")

    location = "Not available"
    city = "Not available"

    if location_tag:

        location_span = location_tag.find("span")

        if location_span:

            location_text = location_span.get_text(
                " ",
                strip=True
            )

            # Example:
            # Nungambakkam, Chennai

            if "," in location_text:

                parts = [
                    x.strip()
                    for x in location_text.split(",", 1)
                ]

                location = parts[0]
                city = parts[1]

            else:
                location = location_text
                city = "Chennai"


    # -----------------------------
    # IMAGE URL
    # -----------------------------

    image_tag = car.find(
        "img",
        class_="hover 3"
    )

    if image_tag:

        image_url = image_tag.get("src")

        # If src is not available, try other attributes
        if not image_url:
            image_url = image_tag.get("data-src")

        if not image_url:
            image_url = image_tag.get("data-lazy")

        if not image_url:
            image_url = "Not available"

    else:
        image_url = "Not available"


    # -----------------------------
    # ADD DATA TO LIST
    # -----------------------------

    data.append({
        "car_make": car_make,
        "car_model": car_model,
        "car_name": car_name,
        "manufacturing_year": manufacturing_year,
        "price": price,
        "kilometers_driven": kilometers_driven,
        "transmission": transmission,
        "fuel_type": fuel_type,
        "location": location,
        "city": city,
        "image_url": image_url
    })


# 7. Convert to DataFrame
df = pd.DataFrame(data)


# 8. Display results
print("\nScraping completed!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df)


# 9. Save CSV
df.to_csv("used_cars_scraped.csv", index=False)

print("\nCSV saved as used_cars_scraped.csv")
````
